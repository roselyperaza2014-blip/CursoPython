```python
import pandas as pd

def read_csv_to_dataframe(file_path: str) -> pd.DataFrame:
    try:
        df = pd.read_csv(file_path)
        return df
    except Exception as e:
        print(f"Error reading CSV file: {e}")
        return pd.DataFrame()
```

```python
Rosedata_df = read_csv_to_dataframe("Event_core_v4.csv")
```

## Respuesta al ejercicio:
Para mi dataset es mejor usar la prueba de Shapiro-Wilk porque tengo 29 datos y esta prueba es adecuada para muestras pequeñas o moderadas (n < 50 datos, aunque funciona bien hasta n = 2000). La prueba Kolmogorov-Smirnov Test (K-S) es menos sensible que Shapiro-Wilk para muestras pequeñas.

* Prueba Shapiro-Wilk para TOM (Total organic matter):

```python
from scipy.stats import shapiro

TOM = Rosedata_df['total_organic_matter_percent']

stat, p = shapiro(TOM)
print(f'Estadístico= {stat:.5f}, p-valor= {p:.5f}')
if p > 0.05:
    print("La distribución es normal (no se rechaza H0)")
else:
    print("La distribución NO es normal (se rechaza H0)")
```

```python
Shapiro_results_1 = shapiro(TOM)
print(Shapiro_results_1)
```

* Prueba Shapiro-Wilk para Proporción Silt-Clay:

```python
# Prueba Shapiro-Wilk para Proporción Silt-Clay
proporcion_silt_clay = Rosedata_df['proport_silt_clay_percent']

stat, p = shapiro(proporcion_silt_clay)
print(f'Estadístico= {stat:.5f}, p-valor= {p:.5f}')
if p > 0.05:
    print("La distribución es normal (no se rechaza H0)")
else:
    print("La distribución NO es normal (se rechaza H0)")
```

```python
Shapiro_results_2 = shapiro(proporcion_silt_clay)
print(Shapiro_results_2)
```
