```python
import pandas as pd

def read_csv_to_dataframe(file_path: str) -> pd.DataFrame:
    """
    Reads a CSV file and converts it into a pandas DataFrame.
    
    Parameters:
        file_path (str): The path to the CSV file.
        
    Returns:
        pd.DataFrame: A DataFrame containing the CSV data.
    """
    try:
        df = pd.read_csv(file_path)
        return df
    except Exception as e:
        print(f"Error reading CSV file: {e}")
        return pd.DataFrame()
```

```python
Rosedata_df = read_csv_to_dataframe("Event_core_v4.csv")
```

Plot with catplot tipo Violin:

```python
#create your own color array
my_colors = ["#da611b", "#30096e"]

# add color array to set_palette
sns.set(font_scale=2, style="ticks", palette= my_colors)
```

```python
ax = sns.catplot(data=Rosedata_df, 
                x="habitat", y="temperat_grad_c",
                hue="habitat", col="parent_event_id", 
                kind="violin", 
                height=7, aspect=1)

  
ax.set(xlabel='', ylabel='Temperature ($^{0}C$)')

ax.set_titles(col_template="{col_name}", row_template="{row_name}")

if ax._legend is not None:
    plt.setp(ax._legend.get_title(), fontsize=20)

sns.despine(top=False,right=False)

plt.show()
```

Plot three-way ANOVA:

```python
ax = sns.catplot(
    data=Rosedata_df, x="depth_m", 
    y="temperat_grad_c", hue="habitat", col="parent_event_id",
    capsize=.2, palette="YlGnBu_d", errorbar="se",
    kind="point", height=7, aspect=1,
)
ax.despine(left=True)

ax.set(xlabel='', ylabel='Temperature ($^{0}C$)')

ax.set_titles(col_template="{col_name}", row_template="{row_name}")

if ax._legend is not None:
    plt.setp(ax._legend.get_title(), fontsize=20)

sns.despine(top=False,right=False)

plt.show()
```

Scatterplot:

```python
ax = sns.lmplot(x="temperat_grad_c", y="total_organic_matter_percent",
                hue="habitat",col="parent_event_id",
                data=Rosedata_df, 
                height=7, aspect=1)

ax.set(xlabel='Temperature ($^{0}C$)', ylabel=' TOM ($Percent$)')

ax.set_titles(col_template="{col_name}", row_template="{row_name}")

plt.setp(ax._legend.get_title(), fontsize=20)

sns.despine(top=False,right=False)

plt.show()
```
