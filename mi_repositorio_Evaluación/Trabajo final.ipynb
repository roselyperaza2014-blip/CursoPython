### 1. Introduction

Thanatocenosis refers to the assemblage of organismal remains found in a given area without being buried or permanently incorporated into the layers of the lithosphere (Perea et al. 2018). However, in the literature, the term "death assemblages" is more commonly used, which essentially represents the same concept. Kidwell and Tomašových (2013) define death assemblages as the set of taxonomically identifiable organic remains found in the surface mixing layers of landscapes and seafloors.

Marine mollusk death assemblages have been among the most extensively studied, resulting in a vast body of literature on the subject (e.g., Kidwell and Bosence 1991; Kidwell and Flessa 1996; Kowalewski et al. 1998; Grill and Zuschin 2001; Weber and Zuschin 2013; Archuby and Roche 2020; Rojas and Martínez 2020; Chattopadhyay et al. 2021; Steger et al. 2022; Albano et al. 2023; Do Nascimento et al. 2023). The following is a summary of the benefits identified in studies of marine benthic mollusk death assemblages:

* They provide an efficient alternative for estimating species richness due to the high diversity and abundance of shells that can be found in relatively few sediment samples. To study the diversity of a living community, multiple sampling events and large sample sizes are usually required. Owing to time averaging (co-occurrence of shells belonging to several generations), death assemblages often better represent the regional diversity of species than any single living community.
* They have been used to explore diversity in marine habitats such as coral reefs and seagrass beds, as they characterize a type of habitat more effectively than a single census of living species.
* There is high fidelity between living and dead organisms, as it has been observed that most dead shells (80–95%) belong to species also observed alive (Kidwell and Flessa 1996). The greater the proportion of taxa with biomineralized structures (calcite, aragonite, or chitin), the higher the fidelity between the thanatocoenosis and the living community of a given site (Kidwell and Bosence 1991).
* The same sampling and data analysis methods are used as in studies of living communities.
* Large quantities of sediment or the collection of living organisms are not required, making these studies generally less invasive for ecosystems.

Therefore, the thanatocoenosis are a useful tool for assessing mollusk diversity patterns in extensive regions and/or in areas with limited monitoring of the group. The objective of the this project is to evaluate the contribution of death assemblages to the total mollusk community in coral reefs and seagrass beds in the Gulf of Batabanó (GB) and the Gulf of Guanahacabibes (GG), in western Cuba (Figure 1).

### 2. Analysis and Results

I compared the mollusk thanatocoenosis (Peraza-Escarrá et al. 2024) and biocoenosis (reported in Armenteros et al. 2018) from the same sampling units, with the aim of assessing the relative contribution of thanatocoenosis to the abundance and species richness of the total mollusk community. For this purpose, replicates at each site were pooled, and the following formulas were applied:

$$
(1) N\% = \frac{N_{dead}}{N_{dead} + N_{live}} \times 100
$$



$$
(2) S\% = \frac{S_{dead}}{S_{total}} \times 100
$$


Where:
- **N%**: relative contribution of the thanatocoenosis to the total species abundance.  
- **N_dead**: number of dead individuals.  
- **N_live**: number of living individuals.  
- **S%**: relative contribution of the thanatocoenosis to the total species richness.  
- **S_total**: species richness including both living and dead individuals.  
- **S_dead**: species richness of the thanatocoenosis.

To show N(%) y S (%) per habitat in each gulf:

```python
import pandas as pd
from tabulate import tabulate

# Create the DataFrame with the provided data
data = {
    "Site": ["GB09","GB10","GB11","GB12","GG01","GG02","GG04","GG05","GG06","GG09","GG14"],
    "Habitat": ["coral_reef","seagrass","coral_reef","seagrass","coral_reef","coral_reef","coral_reef","seagrass","seagrass","seagrass","seagrass"],
    "N_live": [5,10,23,6,10,1,17,13,17,26,20],
    "N_dead": [1553,744,1118,355,364,139,660,752,576,474,378],
    "S_live": [5,5,6,4,8,1,6,6,5,9,9],
    "S_dead": [181,79,166,55,83,40,142,83,58,71,68],
    "S_total": [182,79,170,58,86,40,145,86,61,78,73]
}

df = pd.DataFrame(data)

# Calculating N% y S%
df["N_percent"] = df["N_dead"] / (df["N_dead"] + df["N_live"]) * 100
df["S_percent"] = df["S_dead"] / df["S_total"] * 100

# Show the DataFrame as a formatted table
print(tabulate(df, headers='keys', tablefmt='github', showindex=False))

# Save the DataFrame to a CSV file
df.to_csv("mollusk_data_analysis.csv", index=False)
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Site": ["GB09","GB10","GB11","GB12","GG01","GG02","GG04","GG05","GG06","GG09","GG14"],
    "Habitat": ["coral_reef","seagrass","coral_reef","seagrass","coral_reef","coral_reef","coral_reef","seagrass","seagrass","seagrass","seagrass"],
    "Gulf": ["Batabano","Batabano","Batabano","Batabano","Guanahacabibes","Guanahacabibes","Guanahacabibes","Guanahacabibes","Guanahacabibes","Guanahacabibes","Guanahacabibes"],
    "N_live": [5,10,23,6,10,1,17,13,17,26,20],
    "N_dead": [1553,744,1118,355,364,139,660,752,576,474,378],
    "S_live": [5,5,6,4,8,1,6,6,5,9,9],
    "S_dead": [181,79,166,55,83,40,142,83,58,71,68],
    "S_total": [182,79,170,58,86,40,145,86,61,78,73]
}

df = pd.DataFrame(data)

df["N_percent"] = df["N_dead"] / (df["N_dead"] + df["N_live"]) * 100
df["S_percent"] = df["S_dead"] / df["S_total"] * 100

# the graphs
fig, axes = plt.subplots(1, 2, figsize=(14,6))

# N_percent per habitat
sns.barplot(x="Habitat", y="N_percent", hue="Gulf", data=df, ax=axes[0])
axes[0].set_title("N% por Habitat y Golfo")
axes[0].set_ylabel("N%")
axes[0].set_ylim(0, 105)

# S_percent per habitat
sns.barplot(x="Habitat", y="S_percent", hue="Gulf", data=df, ax=axes[1])
axes[1].set_title("S% por Habitat y Golfo")
axes[1].set_ylabel("S%")
axes[1].set_ylim(0, 105)

plt.tight_layout()
plt.show()
```

The results showed the high contribution (> 90%) of death assemblages to the abundance and species richness of the total mollusk community. It was confirmed that death assemblages have higher species richness than any single census of living species, which is explained by the accumulation of shells from different generations over time (time averaging) (Kidwell and Bosence 1991).

Many species found exclusively in the thanatocoenosis are not allochthonous or exotic, but have simply not yet been collected alive, possibly due to their aggregated distribution or the ephemeral nature of their populations (Kidwell 2008). On the other hand, some species from the biocoenosis did not occur in the thanatocoenosis at the same site, but were observed in the thanatocoenosis of other sites within the same gulf. It was also observed that a few species exclusive to the biocoenosis of one gulf were present in the thanatocoenosis of the other gulf (e.g., *Angulus* sp., *Atys* spp., *Caecum* spp., and *Phlyctiderma semiasperum*). All of this indicates that species “absent” in the local thanatocoenosis were present in the regional species pool.
